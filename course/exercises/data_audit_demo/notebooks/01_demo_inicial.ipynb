# Demo inicial: Juntos y asistencia escolar

**Datos sintéticos.** Pregunta descriptiva: ¿qué patrones muestran estos datos sobre asistencia escolar entre estudiantes de hogares que participan y no participan en Juntos?

> Este ejercicio no identifica efectos causales.

```python
from pathlib import Path
import pandas as pd

CANDIDATES = [
    Path.cwd(),
    Path.cwd() / 'course/exercises/data_audit_demo',
    Path.cwd().parent,
]
ROOT = next(
    (p for p in CANDIDATES if (p / 'data/estudiantes.csv').exists()),
    None,
)
if ROOT is None:
    raise FileNotFoundError('Abre la carpeta data_audit_demo o la raíz del repositorio')
DATA = ROOT / 'data'
print('Raíz del demo:', ROOT.resolve())
```

```python
estudiantes = pd.read_csv(DATA / 'estudiantes.csv')
hogares = pd.read_csv(DATA / 'caracteristicas_hogar.csv')

estudiantes.head()
```

## Construcción de la base analítica

```python
analisis = estudiantes.merge(
    hogares,
    on='hogar_id',
    how='inner',
)

print(f'Base analítica: {len(analisis):,} estudiantes')
```

```python
tabla_inicial = (
    analisis.groupby('juntos', as_index=False)
    .agg(
        estudiantes=('estudiante_id', 'size'),
        asistencia_media=('asiste_regularmente', 'mean'),
    )
)
tabla_inicial['asistencia_media'] = tabla_inicial['asistencia_media'].round(3)
tabla_inicial
```
